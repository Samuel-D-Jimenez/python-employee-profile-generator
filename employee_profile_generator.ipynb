```python
# Define the function
def generate_employee_cards():
  print("=== Welcome to the Employee Profile Generator ===")

  # Ask how many employees to generate using a loop
  try:
    num_employees = int(
        input("How many employee profiles would you like to create? ")
    )
  except ValueError:
    print("Please enter a valid number.")
    return

  employee_list = []

  for i in range(num_employees):
    print(f"\n--- Entering details for Employee {i + 1} ---")
    first_name = input("First Name: ").strip()
    last_name = input("Last Name: ").strip()
    employee_age = int(input("Age: "))
    position = input("Position / Job Title: ").strip()
    salary = float(input("Salary (e.g., 75000): "))
    employee_code = input(
        "Employee ID Code (format: e.g., DEV-2026-JD-001): "
    ).strip()

    # Processing data using f-strings and slicing
    full_name = f"{first_name} {last_name}"
    department = employee_code[:3]
    year_code = employee_code[4:8]
    initials = employee_code[9:11]

    # Store card summary in a dictionary/list
    card_info = {
        "card": (
            f"Employee: {full_name} | Age: {employee_age} | Position:"
            f" {position} | Salary: ${salary:,.2f}"
        ),
        "metadata": (
            f"  -> Dept: {department} | Year: {year_code} | Initials:"
            f" {initials}"
        ),
    }
    employee_list.append(card_info)

  # Display all generated cards
  print("\n========================================")
  print("       GENERATED EMPLOYEE CARDS         ")
  print("========================================")
  for index, emp in enumerate(employee_list, start=1):
    print(f"\n[Card #{index}]")
    print(emp["card"])
    print(emp["metadata"])
```

```python
# Run the function
generate_employee_cards()
```
